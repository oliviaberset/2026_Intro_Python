## Exercise 2: Unit conversions

A temperature is recorded as 18.5 °C. Compute it in kelvin and in degrees Fahrenheit. Separately, a discharge of 4.7 m³ s⁻¹ drains a catchment of area 250 km². Express the specific discharge in mm per day (use: 1 m³ s⁻¹ spread over 1 km² for one day equals 86.4 mm). Print each result to two decimals with an f-string.

````python
# Your solution here
temperature_celsius = 18.5
temperature_kelvin = temperature_celsius + 273.15
temperature_fahrenheit = (temperature_celsius * 1.8) + 32

discharge = 4.7
area = 250

discharge_mm_day = (discharge * 86.4) / area

print(f"Temperature in kevlin: {temperature_kelvin: .2f} K")
print(f"Temperature in fahrenheit: {temperature_fahrenheit: .2f}°F")

print(f"Discharge: {discharge_mm_day: .2f} mm/day")

````

## Exercise 3: Operator families and a quality flag

Three daily readings arrive from a station: `reading_1_celsius` = 18.2, `reading_2_celsius` = -300.0, and `reading_3_celsius` = None (the sensor reported nothing).

1. Build a boolean `is_plausible` that is `True` when the first reading is above absolute zero and below 60 °C.
2. Build `is_implausible` for the second reading: apply the same test as in step 1, and negate it with `not`.
3. Use `is` to check whether the third reading is missing.
4. Starting from `n_valid = 0`, use `+=` twice to count how many of the first two readings pass the step 1 test. Note that `int()` casts `True` to 1 and `False` to 0.
5. The sensor is rated for -40 °C to 60 °C, both bounds included. Use `>=` and `<=` to test whether the first reading lies in that range, and `!=` to check that the first two readings differ.

````python
# Your solution here
reading_1_celsius = 18.2
reading_2_celsius = -300.0
reading_3_celsius = None

# Question 1
is_plausible = 60.0 > reading_1_celsius > -273.15
print(is_plausible)

# Question 2
is_implausible = not(60.0 > reading_2_celsius > -273.15 )
print(is_implausible)

# Question 3
reading_3_celsius is None

# Question 4
n_valid = 0
pass_step1 = int(is_plausible) + int(not(is_implausible))
n_valid += pass_step1
print(n_valid)

# Question 5
lies_in_range = -40.0 <= reading_1_celsius <= 60
print(lies_in_range)

reading_1_celsius != reading_2_celsius
````

## Exercise 4: Parse a record

A logger emits records like `"2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"`. Extract the date string, the station name in title case, the numeric value as a float, and the unit as written, then print them. Then print the station name in lower case as well, the form you would use in a file name.

````python
# Your solution here
records = "2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"
records_split = records.split(",")

date = records_split[0].strip()
station_name = records_split[1].strip().title()
value = float(records_split[2].strip())
unit = records_split[3].strip()

print(date, station_name, value, unit)

print(station_name.lower())
````

## Exercise 7: Round-trip through a file, with an append

Write the Jungfraujoch week from the lecture — the daily maximum temperatures -8.4, -8.9, -9.2,
-9.0, -9.5 (°C) — to `_files/jfj_week.txt`, one value per line, using `pathlib` and a `with`
block. Create the `_files` folder first if it does not exist; do not rely on Exercise 6 having
made it.

1. Write the five values with mode `"w"`, one `.write()` call per value.
2. Reopen the file with mode `"a"` and append the last two days of the week, `-7.8` and `-6.1`.
3. Read the file back with `.readlines()`, cast each of the seven lines to `float` by name, and
   compute the mean by summing the seven values by hand with `+` and dividing by `len()`. Print
   it to two decimals.

````python
# Your solution here
from pathlib import Path

data_path = Path("_files")
data_path.mkdir(exist_ok=True)

file_path = data_path/"jfj_week.txt"

with file_path.open("w", encoding="utf-8") as fhandler:
  fhandler.write("-8.4\n")
  fhandler.write("-8.9\n")
  fhandler.write("-9.2\n")
  fhandler.write("-9.0\n")
  fhandler.write("-9.5\n")

with file_path.open("a", encoding="utf-8") as fhandler:
  fhandler.write("-7.8\n")
  fhandler.write("-6.1\n")

with file_path.open("r", encoding="utf-8") as fhandler:
  all_lines = fhandler.readlines()
for line in all_lines:
  values = float(line)
  print(values)

values = [ -8.4, -8.9, -9.2, -9.0, -9.5, -7.8, -6.1]
sum_values = values[0] + values[1] + values[2] + values[3] + values[4] + values[5] + values[6]
print(f"Moyenne:", round(sum_values / len(values),2))
````

````python
# Your solution here
import math

# Question 1
temperature = [-8.4, -8.9, -9.2, -9.0, -9.5, -7.8, -6.1]
mean_temp = (values[0] + values[1] + values[2] + values[3] + values[4] + values[5] + values[6])/len(temperature)
print(mean_temp)

# Question 2
mean_temp_reverse = (values[6] + values[5] + values[4] + values[3] + values[2] + values[1] + values[0])/len(temperature)
print(mean_temp_reverse)

# Question 3
mean_temp == mean_temp_reverse

# Question 4
print(math.isclose(mean_temp, mean_temp_reverse))
````
