## Exercise 1: Build a GeoDataFrame

Create a GeoDataFrame of two points, `A` at (7.4, 46.9) and `B` at (8.5, 47.4), in EPSG:4326. Print the epsg code and the geometry column.

````python
# Your solution here

import geopandas as gpd
import pandas as pd

data_df = pd.DataFrame({
    "A":[7.4, 46.9],
    "B":[8.5, 47.4],
})

data = gpd.GeoDataFrame(
    data_df,
    geometry=gpd.points_from_xy(data_df["A"], data_df["B"]),
    crs="EPSG:4326",
)

print(f"epsg code: {data.crs.to_epsg()}")
data["geometry"]
````

## Exercise 2: Write and read GeoJSON

````python
# Your solution here
from pathlib import Path

data_path = Path("_files/points.geojson")
data_path.parent.mkdir(exist_ok=True)

gpd.GeoDataFrame(
    data_df,
    geometry=gpd.points_from_xy(data_df["A"], data_df["B"]),
    crs="EPSG:4326",
).to_file(data_path, driver="GeoJSON")

with open(data_path, "r", encoding="utf-8") as f:
  read = f.read()
print(read)
````
